## 2. The question: decision, action, cost of a wrong call

**Search question:** For a content strategist managing many pages, which pages should enter the next refresh queue first to reduce near-term decline risk and recover search opportunity?

**Decision and action:** The strategist uses a ranked list to pick the next batch of pages to audit and update (title/intent alignment, content depth, and freshness updates).

**Cost of wrong recommendations:** A false positive wastes editor time on pages that did not need intervention; a false negative delays updates on pages that are actually slipping, which can mean missed clicks/sessions and slower recovery.

```python
decision_owner = 'Content strategist / SEO editor'
action = 'Review and refresh highest-priority pages first'
error_costs = {
    'false_positive': 'Wasted editorial time',
    'false_negative': 'Missed recovery window on declining pages'
}
print('Decision owner:', decision_owner)
print('Action:', action)
print('Error costs:', error_costs)
```

## 3. Quick look at the data (2-3 real numbers)

A quick starter-data scan supports this lane:

1. The file has **30,000 pages across 32 clients**, so there is enough scale for prioritization choices.
2. **9,961 pages (33.2%)** are already marked `trend_direction = down` with at least 500 impressions in 90 days, so the decline problem is not rare.
3. **9,759 pages (32.5%)** are visible (`impressions_90d >= 500` and `0 < avg_position <= 20`) but have low CTR (`ctr < 0.5`, where 0.5 means 0.5%), which signals a sizeable optimization queue.

These numbers suggest the capstone can improve a real triage decision, not just run a model for its own sake.

```python
import csv
from pathlib import Path

csv_path = Path('../../data/raw/content_refresh_anonymized.csv')
rows = list(csv.DictReader(csv_path.open()))

def to_float(v):
    try:
        return float(v)
    except Exception:
        return float('nan')

total_rows = len(rows)
n_clients = len({r['client_id'] for r in rows})
down_visible = sum((r['trend_direction'] == 'down') and (to_float(r['impressions_90d']) >= 500) for r in rows)
low_ctr_visible = sum((to_float(r['impressions_90d']) >= 500) and (0 < to_float(r['avg_position']) <= 20) and (to_float(r['ctr']) < 0.5) for r in rows)

print(f'Rows: {total_rows:,}')
print(f'Clients: {n_clients}')
print(f'Down + >=500 impressions: {down_visible:,} ({down_visible/total_rows:.1%})')
print(f'Visible but low CTR: {low_ctr_visible:,} ({low_ctr_visible/total_rows:.1%})')
```
